## **Problem Statement**
Consider a finite, simply-connected open domain $\Omega \subset \mathbb{R}^{2}$ with a smooth boundary $\partial\Omega$ and the outward normal vector $\mathbf{n}$. ### **Question 1: Global Conservation Law & Dimensional Analysis (2 p)**
Denoting by $c(x,y,t)$ $[\text{m}^{-2}]$ the concentration of the substance and by $\mathbf{J}(x,y,t)$ its flux density, write down the global conservation law. What are the physical units of $f$ and $\mathbf{J}$?

*Type your answer and unit derivation below:* 






Let $c(x,y,t)$ be the concentration of the substance, and let $\mathbf{J}(x,y,t)$ be its flux density. 
The total amount of substance inside $\Omega$ at time $t$ is

$$ Q(t)=\iint_{\Omega} c(x,y,t) dA $$

The global conservation law is

$$ \frac{dQ}{dt}=-\Phi+R $$

where $\Phi$ is the outward flux through the boundary $S$ and $R$ is the speed of creation/destruction of the substance inside $\Omega$. Hence,

$$ \Phi=\int_{S}\mathbf{J}\cdot\mathbf{n}ds $$

and

$$ R=\iint_{\Omega}f dA. $$

Therefore, for the two-dimensional problem,

$$ \frac{d}{dt} \iint_{\Omega}cdA = -\int_{S}\mathbf{J} \cdot \mathbf{n}ds + \iint_{\Omega}fdA $$

This conservation law must hold for every subdomain $A$ inside $\Omega$.

Since $ [c]= m^{-2}, $ we have $ \left[\frac{\partial c}{\partial t}\right] = m^{-2}s^{-1}. $ Therefore, $ [f]=\mathrm{m}^{-2}\mathrm{s}^{-1}.$
For the flux density $\mathbf{J}$, the total amount of substance has units $ [Q] = [c][dA] = m^{-2}m^{2} = 1. $ Hence, $ \left[\frac{dQ}{dt}\right] = s^{-1}. $
The outward flux through the boundary is $ \Phi=\int_S\mathbf{J}\cdot\mathbf{n}ds. $
Since $[ds]=m$ and $n$ is dimensionless, we require $ [\mathbf{J}][ds] = s^{-1}. $
Therefore, $ [\mathbf{J}] = \frac{s^{-1}}{m} = m^{-1}s^{-1} $
Thus, $ [f]=m^{-2}s^{-1}, [J]=m^{-1}s^{-1} $.

### **Question 2: Local Conservation Law Derivation (2 p)**
Derive the local conservation law for $(x,y) \in \Omega$ by applying the Divergence Theorem to your global conservation law under the assumption that it holds for any sub-domain of $\Omega$ for all $t > 0$.

*Type your step-by-step derivation below:* 

Starting from the global conservation law,

$$ \frac{d}{dt}\iint_{\Omega}cdA = -\int_{S}\mathbf{J}\cdot\mathbf{n}ds + \iint_{\Omega}fdA $$

Since the domain $\Omega$ is fixed in time, the time derivative can be moved inside the integral:

$$ \iint_{\Omega}\frac{\partial c}{\partial t}dA = -\int_{S}\mathbf{J}\cdot\mathbf{n}ds + \iint_{\Omega}fdA $$

Using the divergence theorem in two dimensions,

$$ \int_{S}\mathbf{J}\cdot\mathbf{n}ds = \iint_{\Omega}\nabla\cdot\mathbf{J}dA $$

Therefore,

$$ \iint_{\Omega}\frac{\partial c}{\partial t}dA = -\iint_{\Omega}\nabla\cdot\mathbf{J}dA + \iint_{\Omega}fdA $$

Collecting all terms into one integral gives

$$ \iint_{\Omega} \left( \frac{\partial c}{\partial t} + \nabla\cdot\mathbf{J} - f \right) dA = 0 $$

Since the conservation law must hold for every subdomain of $\Omega$, the integrand itself must vanish. Hence, the local conservation law is

$$ \frac{\partial c}{\partial t} + \nabla\cdot\mathbf{J} = f , (x,y)\in\Omega $$

In two dimensions,

$$ \nabla\cdot\mathbf{J} = \frac{\partial J_x}{\partial x} + \frac{\partial J_y}{\partial y}, $$

so the local conservation law can also be written as

$$ \frac{\partial c}{\partial t} + \frac{\partial J_x}{\partial x} +\frac{\partial J_y}{\partial y} = f $$

From the derived governing equation,

$$ \frac{\partial c}{\partial t} - \nabla\cdot(k\nabla c) = f, (x,y)\in\Omega, $$

since we are looking at the steady-state BV problem we can reduce $\partial c/\partial t = 0$. Hence, the full steady-state BV problem can be stated as, 

$$ -\nabla \cdot (k\nabla c) = f, (x,y)\in \Omega $$

$$ k\frac{\partial c}{\partial n}+\alpha c=0,(x,y)\in\partial\Omega. $$

(a)
Since in the boundary condition we have both c and its derivative (Dirichlet + Neumann), and the right-hand side of the condition is zero, the BC is a homogeneous Robin BC.

For classification, we can expand the PDE as,

$$ -\nabla \cdot (k\nabla c) = -(kc_x)_x - (kc_y)_y = -kc_{xx} - kc_{yy} - k_xc_x-k_yc_y = f $$

The second order part is what's needed for classification. The PDE can be written down as,

$$\nabla \cdot A \nabla u + F_{first order} = f$$

With the matrix $A = \begin{pmatrix} k & 0 \\ 0 & k \end{pmatrix} = kI $. 
The characteristic equation becomes $(k-\lambda)^2 = 0$ and therefore the eigenvalues of the matrix are both $\lambda_1 = k$ and $\lambda_2 = k$. Since $k(x,y) > 0$, both eigenvalues are nonzero and have the same sign. This corresponds with an elliptic PDE. It is a linear second-order ellipc PDE.

For $k = 1$, the steady-state problem becomes
$$ -(\Delta c) = f, (x,y)\in \Omega $$

$$ k\frac{\partial c}{\partial n}+\alpha c=0,(x,y)\in\partial\Omega. $$

Theorem: it can be proven that the Robin problem above consisting of $\Delta u = f$ in $\Omega$ and $\sigma u + D_n u = g$ on $\partial \Omega, \sigma>0$ has at most one solution.

Proof:
Let $u_1$ and $u_2$ be two solutions, and define

$$
w = u_1-u_2.
$$

Subtracting the PDEs gives

$$
-\Delta w
=
-\Delta u_1+\Delta u_2
=
f-f
=
0.
$$

Similarly, subtracting the boundary conditions gives

$$
D_n w+\sigma w=0,
\qquad (x,y)\in\partial\Omega.
$$

Since $-\Delta w=0$, $w$ is harmonic and therefore there is a straight line moving from boundary to boundary. Therefore, by the maximum and minimum principles, the maximum and minimum of $w$ on $\overline{\Omega}$ are
attained on $\partial\Omega$, unless $w$ is constant.

Suppose first that $w$ has a positive maximum $M$ inside $\overline{\Omega}$. Let $(x_0,y_0) \in\partial\Omega$ be a point where this maximum is attained.
Since $\mathbf{n}$ is the outward normal, moving in the direction of the normal toward the maximum must be increasing, therefore at a boundary maximum we have

$$ D_n w(x_0,y_0)\geq 0. $$

However, from the Robin boundary condition,

$$ D_n w=-\sigma w. $$

Hence, at $(x_0,y_0)$,

$$ D_n w(x_0,y_0) = -\sigma(x_0,y_0)M. $$

Since $\sigma>0$ and $M>0$, $ D_n w(x_0,y_0)<0, $ which is a contradiction. Therefore, $ \max_{\overline{\Omega}}w\leq 0. $

Now suppose that $w$ has a negative minimum $m$ in $\overline{\Omega}.$

Let $(x_1,y_1)\in\partial\Omega$ be a point where this minimum is attained.
At a boundary minimum,

$$ D_n w(x_1,y_1)\leq 0 $$

But the Robin boundary condition gives

$$ D_n w(x_1,y_1) = -\sigma(x_1,y_1)m $$

Since $\sigma>0$ and $m<0$, $ D_n w(x_1,y_1)>0 $ which is again a contradiction. Therefore, $ \min_{\overline{\Omega}}w\geq 0 $

Combining both results,

$$ \max_{\overline{\Omega}}w\leq 0, \min_{\overline{\Omega}}w\geq 0.
$$

Thus, $w=0$ in $\overline{\Omega}$.

Since $w=u_1-u_2$, it follows that $ u_1=u_2.$ Hence, there is at most one solution to the Robin BV problem.
